```python
from langchain_community.document_loaders import PyPDFLoader
pdf_path = r'C:\Users\devad\OneDrive\Desktop\Agentic AI\data\Beauty_dataset.pdf'
loader = PyPDFLoader(pdf_path)
documents = loader.load()
documents
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size = 200,
    chunk_overlap = 3
)

# Extract the text content from each document and join them into a single string
full_text = " ".join([doc.page_content for doc in documents])

chunks = splitter.split_text(full_text)
len(chunks)
```

```python
! pip install langchain-community langchain-huggingface
```

```python
from langchain_community.vectorstores import FAISS
from langchain_huggingface.embeddings import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings()
db = FAISS.from_texts(chunks, embedding_model)
```

```python
chunks
```

```python
results = db.similarity_search("How can the beauty of nature be observed?")
results1 = db.similarity_search("What is Beauty?")
print(results[0].page_content)
print(results1[0].page_content)
```
