## Import the required libraries

```python
from diffusers import StableDiffusionPipeline
import torch
import matplotlib.pyplot as plt
```

## Load the stable diffusion model from hugging face

```python
pipe = StableDiffusionPipeline.from_pretrained("runwayml/stable-diffusion-v1-5", torch_dtype=torch.float16).to("cuda")
```

## Prompt for generation

```python
prompt = "A misty mountain landscape during a light drizzle, dense fog covering the hills, a small cozy tea stall beside a winding road, a motorcycle parked nearby, a bike rider in a rain jacket holding a steaming cup of tea, peaceful atmosphere, wet roads, cinematic mood, realistic photography, soft natural lighting, highly detailed."
```

## Generate Image

```python
with torch.autocast("cuda"):
  image = pipe(prompt).images[0]
```

```python
plt.imshow(image)
plt.axis("off")
plt.title("Generated by stable diffusion")
plt.show()
```

```python
image.save("output_image.png")
```
