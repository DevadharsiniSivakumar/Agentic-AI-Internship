```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression 
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error, mean_absolute_percentage_error
```

## Gathering all required data into dataframes

```python
df = pd.read_csv("C:/Users/devad/OneDrive/Desktop/Agentic AI/salary_dataset.csv")
```

## Defining the Features(X) ad Target(Y) for my data

```python
X = df.drop(['Salary', 'Unnamed: 0'], axis=1) ## Features 
Y = df["Salary"] ## target variable
```

## Splitting the dataset into training and testing part

X_train, X_test, Y_train, Y_test = train_test_split(x,y)

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.2, random_state=42)
```

```python
X_train.shape
```

```python
X_test.shape
```

## Model Selection

```python
model = LinearRegression()
```

## Train Linear Regression Model

```python
model = model.fit(X_train, Y_train)
```

## Predicting Values from the model

```python
predictedSalary = model.predict(X_test)
```

```python
predictedSalary
```

## Comparing predicted value with the og values(Y_test)

```python
errors = mean_squared_error(Y_test, predictedSalary)
print("Mean Squared Error: ", errors)

r2 = r2_score(Y_test, predictedSalary)
print("R2 Score: ", r2)

mae = mean_absolute_error(Y_test, predictedSalary)
print("Mean Absolute Error : ", mae)

mape = mean_absolute_percentage_error(Y_test, predictedSalary)
print("Mean Absolute Percentage Error: ", mape)
```

## Visualising estimated best fit line for the model

```python
plt.scatter(X,Y, color='blue')
plt.plot(X, model.predict(X), color='red')
```
