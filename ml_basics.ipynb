### Step 1. Importing required libraries

```python
## linear algebra
import numpy as np
## data processing, data transformation, data cleaning, csv file I/O
import pandas as pd
## used to perform linear regression
from sklearn.linear_model import LinearRegression 
## used to plot the data
import matplotlib.pyplot as plt
## all these are not required for linear regression
#from sklearn.model_selection import train_test_split
#import tensorflow as tf
#import seaborn as sea
#import plotly as py
```

### Experimenting with random data

```python
%matplotlib inline
plt.style.use('ggplot')
```

```python
randomdata = np.random.RandomState(1)
x = 10 * randomdata.rand(50) ## 50 random numbers btwn 0 to 50
y = x-1+randomdata.rand(50) ## y is a linear function
plt.scatter(x,y)
plt.show()
```

```python
model = LinearRegression()
model.fit(x[:, np.newaxis], y) ## fit the model to the data

xfit = np.linspace(0, 10, 100)
yfit = model.predict(xfit[:, np.newaxis])
plt.scatter(x, y)
plt.plot(xfit, yfit, color='green')
plt.show()
```
